```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xx_data = np.array([0, 0.25, 0.5, 1, 1.5, 2, 2.5, 3])
y_data = np.array([0, 0.06, 0.11, 0.16, 0.27, 0.33, 0.42, 0.46])
x_data = xx_data/5
```

```python
m, c = np.polyfit(x_data, y_data, 1)
y_un = 0.19
x_un = (y_un - c) / m
y_pred = m * x_data + c
```

```python
plt.plot(x_data, y_pred, 'o--', label='Theoretical Line')
plt.title('Protein Estimation Standard Curve')
plt.xlabel('Protein Concentration (mg/mL)')
plt.ylabel('Absorbance (y)')
plt.scatter(x_data, y_data, color='red', label='Lab Data Point')
plt.scatter(x_un, y_un, color='green', marker='X', label=f'Unknown Sample: ({x_un:.2f}, {y_un:.2f})')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
```
